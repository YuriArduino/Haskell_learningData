```haskell
-- primeiro programa em haskell

polinomio :: Int -> Int

polinomio x = x*x +10*x + 2
```

```haskell
-- calcula o quadrado de um inteiro

quadrado :: Int -> Int
quadrado n = n*n

-- soma

soma :: Float -> Float ->Float
soma a b = a + b

--triplica
triplica :: Int -> Int
triplica var = var*3
```

```haskell
quadrado 12

soma 10 20

soma 17.998 1.543

triplica 27
```

```haskell
area :: Float -> Float
area raio = raio*raio*pi

perimetroRaio :: Float -> Float
perimetroRaio raio = 2*pi*raio
```

```haskell
area 10
perimetroRaio 10
```

```haskell
hipotenusa :: Float -> Float -> Float
hipotenusa a b = sqrt(a*a + b*b) -- () calcular precedência de operadores
```

```haskell
hipotenusa 3 4
```

```haskell
diferenca :: Float -> Float -> Float
diferenca r1 r2 = abs (area r1 - area r2) -- retorna absoluto
```

```haskell
diferenca 10 9
diferenca 10 11
```
