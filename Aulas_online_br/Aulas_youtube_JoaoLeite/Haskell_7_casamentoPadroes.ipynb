```haskell
-- Ao chamar  a função passando um parâmetro,
-- o Hakell vai tentar definir em qual padrão ele se encaixa,
-- testando de cima para baixo os padrões fornecidos na função.

padroes :: Int -> String
padroes 1 = "Um!"
padroes 2 = "Dois!"
padroes 3 = "Tres!"
padroes x = "Nao esta entre 1 e 3"
```

```haskell
padroes :: Int -> String
padroes 1 = "Um!"
padroes 2 = "Dois!"
padroes 3 = "Tres!"
padroes _ = "Nao esta entre 1 e 3" -- O "_" é um curinga
                                   -- significa "qualquer outro valor"

padroes 1
padroes 2
padroes 3
padroes 3000
```

```haskell
padroes2 :: [Int] -> Int
padroes2 [] = 0 -- padrão caso base
padroes2 (_:t) = 1 + padroes2 t
-- padrão recursivo, ignora o primeiro elemento da lista
--e soma 1 ao resultado da chamada recursiva

type Tupla4 = (Int,Int,Int,Int) -- Define tipo 4-upla de inteiros
padroes3 :: Tupla4 -> String
padroes3 (_,_,_,fourth)
    | fourth > 10 = "Maior que 10"
    | otherwise = "Não é maior que 10"
```

```haskell
 -- Retorna a soma de todos os elementos da lista
padroes2 [x | x <- [1 .. 1000], even x] -- List comprehension

-- Retorna se o quarto elemento da 4-upla é maior que 10
padroes3 (100,2,35,40)
```

```haskell
f1 :: Int -> Int -> Int -> Int
f1 x y z
    | (x == 1) = 10
    | (y == 2) = 20
    | (z == 3) = 30
    | otherwise = 0

f1 0 2 0
```

```haskell
-- O padrão coringa pode ser muito usado em funções
-- cujos dados indefinidos (variáveis) ou que não são relevantes
-- para a função, não precisam ser nomeados.

f2 :: Int -> Int -> Int -> Int
f2  1 _ _ = 10
f2  _ 2 _ = 20
f2  _ _ 3 = 30
f2  _ _ _ = 0

f2 0 2 0


-- pode ser uma alternativa para as condicionais if-then-else
-- muito grandes ou para guards.
```

```haskell
opp :: (Int,(Int,Int)) -> Int
opp z = if fst z == 1
    then fst (snd z) + snd (snd z) -- primeira resposta
    else if fst z == 2
        then fst (snd z) - snd (snd z) -- segunda resposta
        else 0 -- resposta padrão
```

```haskell
opp (1,(10,20))

opp (2,(10,20))

opp (3,(10,20))
```

```haskell
opp2 :: (Int,(Int,Int)) -> Int
opp2 (1,(x,y)) = x + y -- primeira resposta
opp2 (2,(x,y)) = x - y -- segunda resposta
opp2 _ = 0 -- resposta padrão
```

```haskell
opp2 (1,(10,20))

opp2 (2,(10,20))

opp2 (3,(10,20))
```
