```haskell
maior :: Int -> Int -> Int
maior a b =
    if a>= b
        then a
        else b
```

```haskell
maior 5 7
maior 50 7
```

```haskell
-- Função usando guards
-- Realiza teste aonde o primeiro True será o output
-- Pode ou não conter a palavra otherwise,
-- como última condição caso nenhuma ser verdadeira

fatorial :: Int -> Int
fatorial n
    | n == 0 = 1
    | n > 0 = n * fatorial (n-1)
```

```haskell
fatorial 5
fatorial 4
fatorial 3
fatorial 6
```

```haskell
maiorg :: Int -> Int -> Int
maiorg a b  | a >= b = a
            | a < b = b
```

```haskell
maiorg 10 15
maiorg 100 15
maiorg 7 7
```

```haskell
maiorh :: Int -> Int -> Int
maiorh a b  | a > b = a
            | a < b = b
            | otherwise = 0
```

```haskell
maiorh 10 15
maiorh 100 15
maiorh 7 7
```

```haskell
ispar :: Int -> Bool
ispar x = if mod x 2 == 0
    then True
    else False
```

```haskell
ispar 10
ispar 101
ispar 287
ispar 286
```

```haskell
charcase :: Char -> String

charcase ch
            | ch >= 'a' && ch <= 'z' = "Minusculo"
            | ch >= 'A' && ch <= 'Z' = "Maiusculo"
            | otherwise = "Desconhecido"
```

```haskell
charcase 'r'
charcase 'P'
charcase '4'
```

```haskell
funcao :: Int -> Int -> Int -> Int

funcao a b c
                | a == 0 = b*b  + 3*c
                | a == 1 = 2*c^2 - 3*c
                | a == 2 = 3*c  - b^2
                | otherwise = 0
```

```haskell
funcao 0 2 3
funcao 1 2 3
funcao 0 0 0
```
