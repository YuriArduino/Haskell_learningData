## Somatória

1.  Escreva uma função que cacule a **somatória** dos elementos de um vetor de inteiros.

```haskell
somat :: [Int] -> Int
somat [] = 0
somat (h:t) = h + somat t
```

2.  Escreva uma função que verifique se uma String **possui o caractere** informado (passado como parâmetro)

```haskell
possuiChar :: [Char] -> Char -> Bool
possuiChar [] ch = False
possuiChar (h:t) ch
    | h == ch =True
    | otherwise = possuiChar t ch
```

```haskell
possuiChar "Joao" 'a'
```

```haskell
possuiChar "Joao" 't'
```

3.  Escreva uma função que busque pelo **maior número** em um vetor de inteiros.

```haskell
maiorElem :: [Int] -> Int
maiorElem [] = -1
maiorElem (h:t) = if h >= maiort then h else maiort
    where maiort = maiorElem  t
```

```haskell
maiorElem [70,4, 999,100,47,10]
```

```haskell
maiorElem [0]
maiorElem []
```

4. Façamos uma função que retorna as raízes resais distintas de uma equação do segundo grau (assumindo que o coeficiente do termo de grau dois nunca seja igual zero) na forma de uma **lista**. A equação é:

$ax² + bx +c = 0$

E o discriminante, geralmente conhecido como "_delta_", é dado por:

$Δ = b² -4ac $

Lembre-se que, se o discriminante for igual a 0, há apensa uma raiz real, caso seja positivo, há duas raízes reais distintas e, se for negativo, não há raízes reais para a equação. **Utilize guardas e definições locais (where ou let)!**

```haskell
raizes :: Float -> Float -> Float -> [Float]

raizes a b c
    | delta < 0 = []
    | delta == 0 = [(-b)/(2 * a)]
    | delta > 0 = [(-b - sqrt delta) / (2 * a),
     (-b + sqrt delta) / (2*a)]
    where
        delta = b * b - 4 * a * c
```

```haskell
raizes 1 (-5) 6
```

```haskell
raizes 1 (-4) 4
```

### Quicksort

Algorítimo ordenação de vetores (ou listas) "Quicksort"

1. **Escolhe** um elemento do vetor (pivô)

2. **Particiona o vetor** onde todos os elementos anteriores ao pivô sejam menores que ele, e os posteriores maiores.

3. **Ordena recursivamente** os vetores de elementos menores e maiores.


```haskell
qsort :: [Int] -> [Int]
qsort [] = []
qsort (h:t) = qsort [y | y <- t, y < h] --menores que pivô
                ++ [h] -- o próprio pivô
                ++ qsort [y | y <- t, y >= h] -- maiores que pivô
```

```haskell
qsort [10,20,6,8,3,15,70,54]
```

```haskell
quickSort :: Ord a => [a] -> [a]
quickSort []     = []
quickSort (x:xs) = quickSort menores ++ [x] ++ quickSort maiores
  where
    menores = [a | a <- xs, a <= x]
    maiores = [a | a <- xs, a > x]
```

```haskell
quickSort [10,20,6,8,3,15,70,54]
```
