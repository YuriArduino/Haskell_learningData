```haskell
-- Operam sobre outras funções, como fst e snd
-- Exemplo de função de mapeamento e filtros

dobra :: Int -> Int
dobra n = n + n

dobraLista :: [Int] -> [Int]
dobraLista [] =[]
dobraLista (prim:outros) = (dobra prim) : dobraLista outros

lista = [1,2,3,4,5,6,7,8,9]
dobraLista lista
```

```haskell
-- usando a função map com a recursividade interna

dobra :: Int -> Int
dobra n = n + n

dobraLista :: [Int] -> [Int]
dobraLista = map dobra

lista = [1,2,3,4,5,6,7,8,9]
dobraLista lista
```

```haskell
--Usando combinação de padrões e funções de alta ordem

dobra :: Int -> Int
dobra n = n + n

quadrado :: Int -> Int
quadrado n = n * n

mapInt :: (Int -> Int) -> [Int] -> [Int]
mapInt _ [] = []
mapInt f (prim:outros) = f prim : mapInt f outros
```

```haskell
mapInt dobra [1,2,3,4,5]
mapInt quadrado [1,2,3,4,5]
let triple x = x * 3 in mapInt triple [1,2,3,4,5]
mapInt (+2) [1,2,3,4,5]
```

```haskell
--filtro usando booleanos

filtro :: (Int -> Bool) -> [Int] -> [Int]
filtro f [] = []
filtro f (h:t)
    | (f h) == True = h : filtro f t
    | otherwise = filtro f t

pares :: Int -> Bool
pares x = (mod x 2) == 0

impares :: Int -> Bool
impares x = (mod x 2) == 1
```

```haskell
filtro pares [1,2,3,4,5,6,7,8,9]
filtro impares [1,2,3,4,5,6,7,8,9]
```

```haskell
maior :: Int -> Int -> Bool
maior a b = if (a > b) then True else False

menor :: Int -> Int -> Bool
menor a b = if (a < b) then True else False

buscaLista :: (Int -> Int -> Bool) -> [Int] -> Int
buscaLista _ []    = -1
buscaLista _ [h]   = h
buscaLista f (h:t) = if f h x then h else x
  where
    x = buscaLista f t
```

```haskell
buscaLista maior [1,2,3,4,5,6,7,8,9]
buscaLista menor [1,2,3,4,5,6,7,8,9]
```

```haskell
-- Usando funções Booleanas nativas do Haskell
-- (>) (<) (==) (<=)(>=)

buscaLista :: (Int -> Int -> Bool) -> [Int] -> Int
buscaLista _ []    = error "Lista vazia"
buscaLista _ [h]   = h
buscaLista f (h:t) = if f h x then h else x
  where x = buscaLista f t


buscaLista (>) [3, 5, 1, 9, 2]
buscaLista (<) [3, 5, 1, 9, 2]
```

```haskell
-- Usando a função filter do Haskell

filter (>5) [3, 5, 1, 9, 2] -- maior que 5

filter (==1) [3, 5, 1, 9, 2] -- igual a 1

filter even [3, 5, 1, 9, 2] -- pares

filter odd [3, 5, 1, 9, 2] -- impares
```
