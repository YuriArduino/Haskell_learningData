```haskell
-- O uso do type na definição da tupla permite a criação de
-- um sinônimo de um tipo.
-- Contudo sua utilização mão cria um novo tipo
-- apenas da um novo nome para um tipo já existente.

type NomeAluno = String
type MediaNota = Int

type Aluno = (NomeAluno, MediaNota) -- Nome, nota
type Turma = [Aluno] -- Lista de alunos
```

```haskell
-- Possibilitando então casamento de padrões

aprovados :: Turma -> Int -> [NomeAluno]
aprovados turma nota = [nome | (nome, media) <- turma,media >= nota]
```

```haskell
turma = [("João", 7), ("Maria", 9), ("José", 5), ("Ana", 8)]
aprovados turma 7
```

1.  Crie a representação de um ponto de três dimensões (x,y,z). A representação deve ser realizada por meio de uma tupla e a definição de um novo tipo.


2. Escreva uma função que calcule a distância entre dois pontos passados como argumentos

```haskell
type Ponto = (Float, Float, Float) -- (x,y,z)

distancia :: Ponto -> Ponto -> Float
distancia (x1, y1,z1) (x2, y2,z2) =
    sqrt (dx^2 + dy^2 + dz^2)
    where
        dx = x1 - x2
        dy = y1 - y2
        dz = z1 - z2
```

```haskell
p1 = (0,0,1)
p2 = (0,0,0)

distancia p1 p2
```

```haskell
distancia (10,5,7) (9,1,8)
```
