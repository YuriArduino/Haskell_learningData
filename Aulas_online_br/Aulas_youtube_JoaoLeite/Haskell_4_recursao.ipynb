```haskell
fatorial :: Integer -> Integer
fatorial n
    | n == 0    = 1                     -- Caso base (solução direta)
    | n > 0     = n * fatorial (n - 1)  -- Caso recursivo (onde ocorre a COMBINAÇÃO da parcial com 'n')

fatorial 5                              -- Chamada da função / Exemplo de execução
```

```haskell
-- divisão recursiva para encontrar o resto, mesma função do mod

divrec :: Int -> Int -> Int
divrec a b
    | b > a = a
    | b == a = 0
    | otherwise = divrec (a-b) b
```

```haskell
divrec 15 4
divrec 15 7
```

```haskell
-- Calcular multiplicação realizando somas sucessivas

multrec :: Int -> Int -> Int
multrec x n
    | x == 1 = n
    | x > 1 = n + multrec (x-1) n
```

```haskell
multrec 10 7
multrec 12 7
multrec 12313 453242
```

```haskell
mdc :: Int -> Int -> Int
mdc x y
    | x > y     = mdc (x - y) y
    | x < y     = mdc y x
    | otherwise = x
```

```haskell
mdc 42 56
```

```haskell
-- Recursão em cauda
-- resultado da chamada recursiva não precisa ser processada
-- para produzir o resultado final
```

```haskell
-- Transformação para recursão em cauda
-- Exemplo a baixo não é recursão em cauda, precisa multiplicar

-- Mudamos para Integer para suportar números gigantescos
potencia2 :: Integer -> Integer
potencia2 n
    | n == 0    = 1
    | otherwise = 2 * potencia2 (n-1)

-- abaixo a versão em cauda
-- ou seja não precisa de cálculo adicional
-- é necessário um segundo paramêtro

potencia2cauda :: Integer -> Integer -> Integer
potencia2cauda n acumulado
    | n == 0 = acumulado
    | n > 0 = potencia2cauda (n-1) (2 * acumulado)
```

```haskell
potencia2 100000
```

```haskell
-- 1 é um valor neutro

potencia2cauda 100000 1
```

```haskell
--Versão melhorada com o operador de avaliação estrita 'seq' ou (!)

-- Usando avaliacao estrita no acumulador para ganho real de performance
potencia2cauda2 :: Integer -> Integer -> Integer
potencia2cauda2 n !acumulado
    | n == 0    = acumulado
    | otherwise = potencia2cauda2 (n-1) (2 * acumulado)
```

```haskell
potencia2cauda2 100000 1

-- Recursão em cauda em Haskell só entrega o ganho real
-- esperado de memória e performance se forçar
-- a computação imediata (usando seq ou !).
```

```haskell
-- fatorial com recursão em cauda

fatorialcauda :: Integer -> Integer -> Integer
fatorialcauda n !acc  -- acc (acumulado)
    | n == 0 = acc
    | n > 0 = fatorialcauda (n-1) (n * acc)
```

```haskell
fatorialcauda 5000 1
```

```haskell
fatorial 5000
```

```haskell
-- calcula o enesimo numero na sequencia de fibonnacci
-- utiliza três expressões de guarda, fechando com "otherwise"

fib :: Int -> Int
fib x
    | x ==0 =1
    | x ==1 =1
    | otherwise = fib(x-1) + fib(x-2)
```

```haskell
fib 10
```

```haskell
fibocauda :: Int -> Int -> Int -> Int
fibocauda n a1 a2
    | n == 0 = a1
    | n == 1 = a2
    | n > 1 = fibocauda (n-1) a2 (a1+a2)
```

```haskell
fibocauda 10 0 1
fibocauda 10 1 1
fibocauda 09 1 1
fibocauda 08 1 1
fibocauda 05 1 1
```
