```haskell
-- usamos 'where' definições locais

areaheron :: Float -> Float -> Float -> Float
areaheron a b c = sqrt (s*(s-a)*(s-b)*(s-c))
    where
        s = (a+b+c)/2
```

```haskell
areaheron 4 5 6
areaheron 20 20 15
```

```haskell
eq2grau :: Float -> Float -> Float -> Int

eq2grau a b c
        | delta > 0 = 2
        | delta == 0 = 1
        | otherwise = 0
    where
        delta = b^2 - 4 * a * c
```

```haskell
eq2grau 1 (-5) 6
eq2grau 1 4 4
eq2grau 1 4 40
```

```haskell
areaCilindro :: Float -> Float -> Float

--calcula a área do cilindro a partir de dois valores:
-- raio da base e -- altura do cilindro
areaCilindro r h = let areaLado = 2 * pi * r * h
                       areaBase = pi * r * r
                       in areaLado + 2*areaBase
```

```haskell
areaCilindro 10 10

areaCilindro 2 10
```

- **where**, definições são colocadas no final, ao contrário de **let** colocadas no início.

- **let** é uma _expressão_ e pode ser usada em qualquer lugar onde se espera uma expressão

> No haskell o próprio corpo de uma função é uma expressão, avaliada por um valor devolvido pela função. A própria construção condicional (if-the-else) é uma expressão, que retorna um valor de cada uma das sua opções. A expressão let/im até parace uma declaração de variável, mas tambem é uma expressçao a ser avaliada por um valor.

- **where** não é uma expressão, podendo ser utilizada apenas para fazer definições locais em uma definição de função.

```haskell
-- agora usamos 'let' definições locais
areaheron2 :: Float -> Float -> Float -> Float
areaheron2 a b c =
    let s = (a + b + c) / 2
    in sqrt (s * (s - a) * (s - b) * (s - c))
```

```haskell
areaheron2 4 5 6
areaheron2 20 20 15
```
